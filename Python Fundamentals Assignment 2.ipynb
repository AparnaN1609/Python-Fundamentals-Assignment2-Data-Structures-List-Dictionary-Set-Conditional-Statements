### 5- Sets- Creation and duplicate values

```python
my_set = set(["a", "e", "i", "o", "u", "a", "a", "i"])
```

```python
print("My Set:", my_set)
```

```python
print("Number of unique elements:", len(my_set))
```

#### Explanation
##### A set stores only unique elements. Duplicate values such as 'a' and 'i' are automatically removed. Sets are unordered, so the displayed order may vary.

#### Attempt to change an element

```python
my_set[4] = "s"
```

#### Explanation 
##### The operation raises a TypeError because sets do not support indexing or item assignment. To add a new element to a set, we can use the add() method.

#### Create two sets

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print("Set 1:", set1)
print("Set 2:", set2)
```

#### Union and Intersection

```python
print("Union:", set1.union(set2))
```

```python
print("Intersection:", set1.intersection(set2))
```

### 6- Operators & Conditional Statements

```python
score = float(input("Enter your score (0 to 10): "))
```

#### Categorize the score

```python
if score < 0 or score > 10:
     print("Invalid score. Please enter a score between 0 and 10.")
elif score > 7:
    print("Above Average: Excellent work! Keep up the good performance.")
elif score >= 4:
    print("Average: Good effort! Keep practicing; there's room for improvement.")
else:
    print("Below Average: Keep practising consistently to improve your performance.")
```
